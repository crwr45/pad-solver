```python
import itertools
from collections import defaultdict
```

```python
solutions = []
with open("../data/data.csv", "r") as f:
    for line in f.readlines():
        solutions.append(line.rstrip().split(','))
```

```python
def all_combinations(l, min, max):
    """all combinations of length min->max for list l"""
    return itertools.chain.from_iterable(
        itertools.combinations(l, i)
        for i in range(min, max + 1))
```

```python
combos_total = defaultdict(lambda : defaultdict(int))
combos_dates = defaultdict(lambda : defaultdict(set))
for sol in solutions:
    date = sol[0]
    sol = sol[1:]
    for c in all_combinations(sol, 1, 6):
        c = frozenset(c)
        combos_total[len(c)][c] += 1
        combos_dates[len(c)][c].add(date)
```

```python
for size, coms in combos_total.items():
    print(size)
    sorted_coms = sorted(coms.keys(), key=coms.get, reverse=True)
    for most_frequent in sorted_coms[:10]:
        print(list(most_frequent), coms[most_frequent])
```

```python
for size, coms in combos_dates.items():
    print(size)
    sorted_coms = sorted(coms.keys(), key=lambda x: len(coms.get(x)), reverse=True)
    for most_frequent in sorted_coms[:10]:
        print(list(most_frequent), len(coms[most_frequent]))
```

```python
# Which dates cannot be solved with "15L0"
days = {}
for sol in solutions:
    if sol[0] not in days:
        days[sol[0]] = False

    if "15L0" in sol:
        days[sol[0]] = True

for d, present in days.items():
    # remove noise from the days covered by "15L0"
    if not present and d[:2] not in ('15', '22', '29', '30', '31'):
        print(d)
```
