```python
from collections import Counter, defaultdict
import math
```

```python
months = ('jan', 'feb', 'mar', 'apr', 'may', 'jun', 'jul', 'aug', 'sep', 'oct', 'nov', 'dec')

solutions = []
with open("../data/data.csv", "r") as f:
    for line in f.readlines():
        solutions.append(line.rstrip().split(','))
```

```python
# Count occurences, and unique dates, for each placement
placement_count = Counter()
dates_by_placement = defaultdict(set)
for s in solutions:
    sol = [a for a in s if a[2:] not in months]
    date = [a for a in s if a[2:] in months][0]
    placement_count.update(sol)
    for p in sol:
        dates_by_placement[p].add(date)
```

```python
# The rarest solution is the one that has the lowest average occurence for all moves.
rarest = ([], math.inf)
for sol in solutions:
    tot = sum([placement_count[p] for p in [a for a in sol if a[2:] not in months]])
    avg = tot / 8
    if avg < rarest[1]:  # lower is rarer
        rarest = (sol, avg)

print(rarest)
```

```python
# Find the solution for each date with the highest average placement occurence
# i.e. the solution for each date comprised of placements which are on average the most common
solution_for_date = defaultdict(lambda: ([], 0))
for s in solutions:
    sol = [a for a in s if a[2:] not in months]
    date = [a for a in s if a[2:] in months][0]
    tot = sum([len(dates_by_placement[p]) for p in sol])
    avg = tot / 8
    if avg > solution_for_date[date][1]:  # higher is more common
        solution_for_date[date] = (sol, avg)

# sort smallest (rarest) to largest (most common)
sorted_dates_by_rarity = sorted(solution_for_date.keys(), key=lambda x: solution_for_date[x][1], reverse=False)
print(sorted_dates_by_rarity[0], solution_for_date[sorted_dates_by_rarity[0]])
```
